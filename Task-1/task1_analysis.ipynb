# Exploratory Data Analysis (EDA) 

## Phase 1: Pandas Statistical Analysis

### ​Q1: Are senior citizens more likely to cancel their subscription than younger customers?


```python
# SeniorCitizen: 0=Not a senior citizen ,1=Senior citizen.
#Churn:No=Customer did not cancel, Yes=Customer cancelled.
# normalize=True gives us the % instead of raw counts, which is more accurate here
senior_churn = df.groupby('SeniorCitizen')['Churn'].value_counts(normalize=True).unstack() * 100

print("Churn Percentage - Senior Citizens:")
print(round(senior_churn, 2))
```

### Q2: Which internet service has the highest churn rate?

```python
# checking churn rates across different internet service types
# using unstack to format the results into a clean grid
internet_churn = df.groupby('InternetService')['Churn'].value_counts(normalize=True).unstack() * 100

print("Internet Service Churn %:")
print(round(internet_churn, 2))
```

### Q3: Do customers with streaming services stay longer with the company?

```python
# separating customers who stream (TV or Movies) from those who don't
has_streaming = df[(df['StreamingTV'] == 'Yes') | (df['StreamingMovies'] == 'Yes')]
no_streaming = df[(df['StreamingTV'] == 'No') & (df['StreamingMovies'] == 'No')]

# comparing average tenure to check if streaming increases loyalty
print("Average tenure (streaming):", round(has_streaming['tenure'].mean(), 1), "months")
print("Average tenure (no streaming):", round(no_streaming['tenure'].mean(), 1), "months")
```

### Q4: Are month-to-month customers more likely to churn than one-year and two-year contract?

```python
# checking how contract length affects churn
# using crosstab to quickly get the row-wise percentages
contract_churn = pd.crosstab(df['Contract'], df['Churn'], normalize='index') * 100

print("Churn % by Contract Type:")
print(round(contract_churn, 2))
```

### Q5: At what tenure are customers most likely to cancel their service?

```python
# isolating customers who actually left
# checking the basic stats (mean, median, etc.) to spot when most drop-offs happen
churned_customers = df[df['Churn'] == 'Yes']

print("Tenure stats for churned customers:")
print(churned_customers['tenure'].describe())
```

### Q6: Does having higher monthly charges increase the likelihood of churn?

```python
# comparing average monthly bills: churned vs retained
charge_impact = df.groupby('Churn')['MonthlyCharges'].mean()

print("Avg monthly charges:")
print(round(charge_impact, 2))
```

### Q7: Which payment method has the highest churn rate?

```python
# checking if payment type (like auto-pay vs electronic checks) impacts churn
payment_churn = df.groupby('PaymentMethod')['Churn'].value_counts(normalize=True).unstack() * 100

print("Churn % by payment method:")
print(round(payment_churn, 2))
```

### Q8: Do customers using multiple services have lower churn than customers using fewer services?

```python
# mapping churn to 1/0 so we can easily calculate the mean
df['Churn_Numeric'] = df['Churn'].map({'Yes': 1, 'No': 0})

# counting how many extra add-on services each customer has
extra_services = ['OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport']
df['ExtraServiceCount'] = (df[extra_services] == 'Yes').sum(axis=1)

# checking if having more services lowers the churn rate
multi_service_churn = df.groupby('ExtraServiceCount')['Churn_Numeric'].mean() * 100

print("Churn % by number of add-on services:")
print(round(multi_service_churn, 2))
```

### Q9: How much total revenue is represented by customers who have churned?

```python
# calculating the total financial hit from churned customers
# using pd.to_numeric defensively to catch any leftover blank spaces in the column
revenue_lost = pd.to_numeric(df[df['Churn'] == 'Yes']['TotalCharges'], errors='coerce').sum()

print("Total revenue lost from churn: $", round(revenue_lost, 2))
```

### Q10: Which customer group has high monthly charges but relatively low churn?

```python
# finding which contract groups have high bills but stay loyal
# .agg() is super handy here to check both metrics at the same time
group_analysis = df.groupby('Contract').agg({
    'MonthlyCharges': 'mean',
    'Churn_Numeric': 'mean' 
})

# formatting the churn column so it reads as a clean % instead of a decimal
group_analysis['Churn_Numeric'] = group_analysis['Churn_Numeric'] * 100
group_analysis.rename(columns={'Churn_Numeric': 'Churn Rate (%)'}, inplace=True)

print("Avg Charge vs Churn by Contract:")
print(round(group_analysis, 2))
```
